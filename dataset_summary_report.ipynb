# Dataset Summary Report

This Colab-ready notebook loads a CSV into Pandas and provides a concise overview of its structure, data quality, and descriptive statistics.

**How to use:** download the CSV from the LMS **Study material** tab, then run the upload cell below.

```python
# Run this cell in Google Colab to upload the CSV downloaded from the LMS.
from google.colab import files
uploaded = files.upload()

csv_files = [name for name in uploaded if name.lower().endswith('.csv')]
if not csv_files:
    raise FileNotFoundError('Please upload a file with a .csv extension.')
csv_path = csv_files[0]
print(f'Using file: {csv_path}')
```

```python
import pandas as pd

# Read the uploaded CSV file into a Pandas DataFrame.
df = pd.read_csv(csv_path)
pd.set_option('display.max_columns', None)
pd.set_option('display.max_colwidth', 80)

print('Dataset loaded successfully.')
```

## 1. Preview the data

```python
print('First five rows')
display(df.head())

print('Last five rows')
display(df.tail())

print('Random sample of five rows')
display(df.sample(min(5, len(df)), random_state=42))
```

## 2. Dataset structure

```python
print(f'Shape (rows, columns): {df.shape}')
print(f'Number of rows: {df.shape[0]}')
print(f'Number of columns: {df.shape[1]}')

print('\nColumn names:')
print(df.columns.tolist())

print('\nIndex:')
print(df.index)

print('\nData types:')
display(df.dtypes.to_frame(name='Data type'))

print('\nDataFrame information:')
df.info()
```

## 3. Descriptive statistics

```python
# include='all' summarizes numeric and non-numeric columns.
display(df.describe(include='all').T)

print('Numeric columns only:')
display(df.describe())
```

## 4. Missing values and observations

```python
missing = df.isna().sum().sort_values(ascending=False)
missing_summary = pd.DataFrame({
    'Missing values': missing,
    'Missing (%)': (missing / len(df) * 100).round(2) if len(df) else 0
})
display(missing_summary)

numeric_columns = df.select_dtypes(include='number').columns.tolist()
non_numeric_columns = df.select_dtypes(exclude='number').columns.tolist()
columns_with_missing = missing[missing > 0]

print('DATASET OBSERVATIONS')
print(f'- The dataset contains {df.shape[0]:,} rows and {df.shape[1]:,} columns.')
feature_names = ', '.join(map(str, df.columns))
numeric_names = ', '.join(map(str, numeric_columns)) or 'None'
non_numeric_names = ', '.join(map(str, non_numeric_columns)) or 'None'
print(f'- Available features: {feature_names}.')
print(f'- Numeric features ({len(numeric_columns)}): {numeric_names}.')
print(f'- Non-numeric features ({len(non_numeric_columns)}): {non_numeric_names}.')
if columns_with_missing.empty:
    print('- No missing values were detected.')
else:
    print(f'- Missing values occur in {len(columns_with_missing)} column(s):')
    for column, count in columns_with_missing.items():
        print(f'  - {column}: {count:,} missing ({count / len(df) * 100:.2f}%)')

print('- Use the descriptive statistics above to identify unusual ranges, skewed numeric values, or dominant categories.')
```
