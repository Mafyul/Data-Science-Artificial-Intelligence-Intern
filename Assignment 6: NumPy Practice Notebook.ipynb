# NumPy for Numerical Data Analysis

This Colab notebook introduces NumPy arrays and the core operations used to explore and analyse numerical data. Run the cells in order; their example outputs are included below each calculation.

## 1. Setup

```python
import numpy as np

print(f'NumPy version: {np.__version__}')
```

## 2. Creating 1D, 2D, and 3D arrays

An array's *axis* represents a dimension. A 1D array is a vector, a 2D array is a table, and a 3D array can represent a stack of tables.

```python
one_d = np.array([10, 20, 30, 40, 50])
two_d = np.array([[1, 2, 3], [4, 5, 6]])
three_d = np.array([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])

print('1D array:\n', one_d)
print('\n2D array:\n', two_d)
print('\n3D array:\n', three_d)
```

## 3. Inspecting dimensions, shape, size, and data type

```python
for name, arr in [('one_d', one_d), ('two_d', two_d), ('three_d', three_d)]:
    print(f'{name}: ndim={arr.ndim}, shape={arr.shape}, size={arr.size}, dtype={arr.dtype}')
```

## 4. Indexing and slicing

NumPy uses zero-based indices. In 2D arrays, use `[row, column]`; slices select a range without changing the original array.

```python
scores = np.array([[72, 88, 91], [65, 79, 84], [90, 94, 87]])
print('Scores:\n', scores)
print('\nFirst row:', scores[0])
print('Last value:', scores[-1, -1])
print('First two rows, last two columns:\n', scores[:2, 1:])
print('Every other element in one_d:', one_d[::2])
```

## 5. Reshaping arrays

`reshape` changes the arrangement of data while preserving the total number of elements.

```python
values = np.arange(1, 13)
matrix = values.reshape(3, 4)
cube = values.reshape(2, 2, 3)

print('Original:', values)
print('\n3 × 4 matrix:\n', matrix)
print('\n2 × 2 × 3 array:\n', cube)
```

## 6. Generating arrays with NumPy functions

```python
print('zeros:\n', np.zeros((2, 3)))
print('\nones:\n', np.ones((2, 2), dtype=int))
print('\narange:', np.arange(0, 10, 2))
print('linspace:', np.linspace(0, 1, 5))
```

## 7. Mathematical and vectorized operations

Vectorized operations apply to whole arrays at once—no Python loop is required.

```python
temperatures_c = np.array([18.0, 21.5, 24.0, 19.5, 22.0])
temperatures_f = temperatures_c * 9 / 5 + 32

print('Celsius:', temperatures_c)
print('Fahrenheit:', temperatures_f)
print('Squared:', temperatures_c ** 2)
print('Square roots:', np.sqrt(temperatures_c))
print('Values above 20°C:', temperatures_c - 20)
```

## 8. Boolean masking

A boolean condition creates a mask that can filter observations meeting a criterion.

```python
sales = np.array([1250, 980, 1430, 760, 1590, 1100])
high_sales = sales >= 1200

print('Sales:', sales)
print('Mask:', high_sales)
print('Sales at least 1200:', sales[high_sales])
print('Count:', high_sales.sum())
```

## 9. Broadcasting

Broadcasting lets NumPy combine compatible shapes. Here, a row of department targets is added to every month of results.

```python
monthly_sales = np.array([[100, 120, 90], [110, 115, 105], [130, 125, 110]])
bonus_by_department = np.array([5, 10, 15])
adjusted_sales = monthly_sales + bonus_by_department

print('Monthly sales:\n', monthly_sales)
print('\nBonus:', bonus_by_department)
print('\nBroadcast result:\n', adjusted_sales)
```

## 10. Aggregate functions

Aggregates summarise a dataset. `axis=0` computes a result per column; `axis=1` computes one per row.

```python
data = np.array([[12, 15, 18], [10, 20, 25], [14, 16, 22]])
print('Data:\n', data)
print('\nTotal:', data.sum())
print('Mean:', data.mean())
print('Minimum:', data.min())
print('Maximum:', data.max())
print('Median:', np.median(data))
print('\nColumn means:', data.mean(axis=0))
print('Row sums:', data.sum(axis=1))
```

## Takeaway

NumPy makes numerical analysis concise and fast: arrays store homogeneous data efficiently, vectorization removes explicit loops, and broadcasting plus aggregation make common analysis tasks natural to express.
